```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick

from src.load import load_case1_data
from src.clean import clean
from src.build_features import build_features, load_config
from src.train import apply_baseline_rule

CONFIG = load_config("../config.yaml")
# Resolve the data location. Participants put the five Kaggle files in
# data/raw/ next to this folder; in the repo they live at the top level.
# Checks for an actual file, so an empty data/raw/ falls through correctly.
def _find_raw_dir(*candidates):
    for c in candidates:
        if os.path.isfile(os.path.join(c, "transactions_data.csv")):
            return c
    raise FileNotFoundError(
        "Could not find transactions_data.csv. Put the five Kaggle files in "
        "data/raw/ (see the .gitkeep there), or edit RAW_DIR below.\n"
        "Looked in: " + ", ".join(candidates)
    )

RAW_DIR = _find_raw_dir("../data/raw", "../../data/raw")
plt.rcParams["figure.dpi"] = 120
```

```python
# Load data — use sample_n=200_000 if you want fast iteration during development.
# Set sample_n=None to train on the full 8.9M-row dataset (~4-5 min total training time).
print("Loading data...")
df_raw = load_case1_data(RAW_DIR, sample_n=200_000, random_state=CONFIG["random_state"])
df = clean(df_raw)
X, y = build_features(df, CONFIG)

print(f"Rows: {len(df):,}")
print(f"Features: {X.shape[1]}")
print(f"Fraud rate: {y.mean():.4%}")
print(f"Feature columns: {list(X.columns)}")
print(y.head())
```
